```python
import os
folder = '/content/pv/raw/color'
for name in sorted(os.listdir(folder)):
    path = os.path.join(folder, name)
    count = len(os.listdir(path))
    print(name, count)
```

```python
import shutil

names = {
    'Tomato___Early_blight': 'Early_blight',
    'Tomato___Late_blight': 'Late_blight',
    'Tomato___Septoria_leaf_spot': 'Septoria_leaf_spot',
    'Tomato___Bacterial_spot': 'Bacterial_spot',
    'Tomato___Leaf_Mold': 'Leaf_mold',
    'Tomato___Tomato_Yellow_Leaf_Curl_Virus': 'Yellow_Leaf_Curl_Virus',
    'Tomato___Tomato_mosaic_virus': 'Mosaic_virus',
    'Tomato___healthy': 'Healthy',
}

drive_folder = '/content/drive/MyDrive/KisaanAI/data/raw/plantvillage'

for old_name, new_name in names.items():
    src = os.path.join(folder, old_name)
    dst = os.path.join(drive_folder, new_name)
    shutil.copytree(src, dst, dirs_exist_ok=True)
    print('Copied', new_name, len(os.listdir(dst)))
```

```python
pd_names = {
    'Tomato Early blight leaf': 'Early_blight',
    'Tomato leaf late blight': 'Late_blight',
    'Tomato Septoria leaf spot': 'Septoria_leaf_spot',
    'Tomato leaf bacterial spot': 'Bacterial_spot',
    'Tomato mold leaf': 'Leaf_mold',
    'Tomato leaf yellow virus': 'Yellow_Leaf_Curl_Virus',
    'Tomato leaf mosaic virus': 'Mosaic_virus',
    'Tomato leaf': 'Healthy',
}

pd_drive = '/content/drive/MyDrive/KisaanAI/data/raw/plantdoc'

for part in ['train', 'test']:
    for old_name, new_name in pd_names.items():
        src = os.path.join('/content/pd', part, old_name)
        dst = os.path.join(pd_drive, part, new_name)
        shutil.copytree(src, dst, dirs_exist_ok=True)
        print('Copied', part, new_name, len(os.listdir(dst)))
```
