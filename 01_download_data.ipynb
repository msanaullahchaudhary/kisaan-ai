```python
raw = '/content/raw'
images = []   # each item: [source, part, class, path]

for cls in sorted(os.listdir(os.path.join(raw, 'plantvillage'))):
    for f in os.listdir(os.path.join(raw, 'plantvillage', cls)):
        images.append(['plantvillage', 'lab', cls, os.path.join(raw, 'plantvillage', cls, f)])

for part in ['train', 'test']:
    for cls in sorted(os.listdir(os.path.join(raw, 'plantdoc', part))):
        for f in os.listdir(os.path.join(raw, 'plantdoc', part, cls)):
            images.append(['plantdoc', part, cls, os.path.join(raw, 'plantdoc', part, cls, f)])

print(len(images), 'images found')
```

```python
from PIL import Image
import imagehash

fingerprints = {}   # fingerprint -> all photos that have it
broken = 0

for i, item in enumerate(images):
    try:
        with Image.open(item[3]) as im:
            h = str(imagehash.phash(im.convert('RGB')))
    except Exception:
        broken += 1
        continue
    fingerprints.setdefault(h, []).append(item)
    if i % 2000 == 0:
        print(i, 'checked')

print('broken:', broken, '  unique photos:', len(fingerprints))
```

```python
keep_first = {'test': 0, 'train': 1, 'lab': 2}
clean = []
duplicates = 0
conflicts = 0

for h, group in fingerprints.items():
    classes = set(item[2] for item in group)
    if len(classes) > 1:            # same photo, different diseases -> drop all copies
        conflicts += len(group)
        continue
    group.sort(key=lambda item: keep_first[item[1]])   # keep a test photo if there is one
    clean.append(group[0])
    duplicates += len(group) - 1

print('duplicates removed:', duplicates)
print('conflicting labels removed:', conflicts)
print('clean images:', len(clean))
```

```python
import random
random.seed(42)

groups = {}
for source, part, cls, path in clean:
    groups.setdefault((part, cls), []).append(path)

rows = []
for (part, cls), paths in sorted(groups.items()):
    paths.sort()
    random.shuffle(paths)
    n = len(paths)
    if part == 'lab':          # PlantVillage: 70% train, 15% validation, 15% test
        a, b = round(n * 0.70), round(n * 0.85)
        parts = [('lab_train', paths[:a]), ('lab_val', paths[a:b]), ('lab_test', paths[b:])]
    elif part == 'train':      # PlantDoc train: 85% train, 15% validation
        a = round(n * 0.85)
        parts = [('field_train', paths[:a]), ('field_val', paths[a:])]
    else:                      # PlantDoc test: only for testing
        parts = [('field_test', paths)]
    for split, ps in parts:
        for p in ps:
            rows.append([p, cls, split])

print(len(rows), 'images given a split')
```

```python
import pandas as pd

df = pd.DataFrame(rows, columns=['path', 'class', 'split'])
df['path'] = df['path'].str.replace(raw + '/', '', regex=False)
df.to_csv('/content/drive/MyDrive/KisaanAI/data/splits.csv', index=False)

pd.crosstab(df['class'], df['split'], margins=True, margins_name='TOTAL')
```
